# Credit Card Fraud Detection — Exploratory Data Analysis

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="darkgrid")
df = pd.read_csv("../data/transactions.csv")
df["transaction_time"] = pd.to_datetime(df["transaction_time"])
df["hour"] = df["transaction_time"].dt.hour
df.head()
```

## Fraud Distribution

```python
plt.figure(figsize=(5,4))
sns.countplot(x="fraud", data=df)
plt.title("Fraud Distribution")
plt.show()
print(df["fraud"].value_counts(normalize=True))
```

## Transaction Amount Histogram

```python
plt.figure(figsize=(7,4))
sns.histplot(df["amount"], bins=50, kde=True)
plt.title("Transaction Amount Distribution")
plt.show()
```

## Fraud vs Amount Boxplot

```python
plt.figure(figsize=(6,4))
sns.boxplot(x="fraud", y="amount", data=df)
plt.ylim(0, df["amount"].quantile(0.99))
plt.title("Amount by Fraud Label")
plt.show()
```

## Fraud by Merchant Category

```python
plt.figure(figsize=(8,4))
rate = df.groupby("merchant_category")["fraud"].mean().sort_values(ascending=False)
sns.barplot(x=rate.index, y=rate.values)
plt.xticks(rotation=30)
plt.ylabel("Fraud Rate")
plt.title("Fraud Rate by Merchant Category")
plt.show()
```

## Fraud by Payment Method

```python
plt.figure(figsize=(7,4))
rate = df.groupby("payment_method")["fraud"].mean().sort_values(ascending=False)
sns.barplot(x=rate.index, y=rate.values)
plt.ylabel("Fraud Rate")
plt.title("Fraud Rate by Payment Method")
plt.show()
```

## Fraud by Location

```python
plt.figure(figsize=(8,4))
rate = df.groupby("location")["fraud"].mean().sort_values(ascending=False)
sns.barplot(x=rate.index, y=rate.values)
plt.xticks(rotation=30)
plt.ylabel("Fraud Rate")
plt.title("Fraud Rate by Location")
plt.show()
```

## Fraud by Age

```python
plt.figure(figsize=(7,4))
sns.histplot(data=df, x="customer_age", hue="fraud", multiple="stack", bins=20)
plt.title("Age Distribution by Fraud Label")
plt.show()
```

## Fraud by Hour

```python
plt.figure(figsize=(8,4))
rate = df.groupby("hour")["fraud"].mean()
sns.lineplot(x=rate.index, y=rate.values, marker="o")
plt.xlabel("Hour of Day")
plt.ylabel("Fraud Rate")
plt.title("Fraud Rate by Hour")
plt.show()
```

## Correlation Heatmap

```python
numeric_df = df[["amount", "customer_age", "previous_transactions", "avg_transaction_amount", "hour", "fraud"]]
plt.figure(figsize=(7,5))
sns.heatmap(numeric_df.corr(), annot=True, cmap="coolwarm", fmt=".2f")
plt.title("Correlation Heatmap")
plt.show()
```
